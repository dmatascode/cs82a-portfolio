## Problem 4.2: Multiplication the Long Way

```python
def multiply_for(a, b):
    result = 0
    for i in range(b):
        result = result + a
    return result

print(multiply_for(7, 6))
```

```python
def multiply_while(a, b):
    result = 0
    count = 0

    while count < b:
        result = result + a
        count = count + 1

    return result

print(multiply_while(7, 6))
```

```python
a = 7
b = 6

print("For loop:", multiply_for(a, b))
print("While loop:", multiply_while(a, b))
print("Regular multiplication:", a * b)
```
