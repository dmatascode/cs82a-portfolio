```python
import pandas as pd

df = pd.read_csv("messy_sales.csv")

print("Shape:", df.shape)

print("\nFirst 10 rows:")
print(df.head(10))

print("\nColumn types:")
print(df.dtypes)
```

## Step 2: First Observations

1. The dates are written in different formats, which could make them harder to analyze.
2. Some products have missing prices shown as NaN.
3. Some ZIP codes have lost their leading zeros because they were stored as numbers.

```python
print(df.isna().sum())
```

```python
median_price = df["price"].median()

df["price"] = df["price"].fillna(median_price)

print("Median price used:", median_price)
print(df.isna().sum())
```

```python
print("Duplicates found:", df.duplicated().sum())

df = df.drop_duplicates()

print("New shape:", df.shape)
```

```python
df["zip"] = df["zip"].astype(str).str.zfill(5)

print(df["zip"].head(10))
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed")

print(df.dtypes)
print(df["date"].head(10))
```

```python
negative_rows = df[df["qty"] < 0]

print(negative_rows)
print("Negative quantity rows:", len(negative_rows))
```

```python
df["is_return"] = df["qty"] < 0

print(df[df["is_return"]])
print("Shape:", df.shape)
```
